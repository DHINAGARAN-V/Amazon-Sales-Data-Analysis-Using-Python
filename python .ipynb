```python
customer.columns = customer.columns.str.strip()
```

```python
marge_data=pd.merge(company,customer,left_on="OrderID",right_on="customer OrderID")
```

```python
#Yearly Revenue
```

```python
yearly_revenue=marge_data.groupby("year")["TotalAmount"].sum()

yearly_revenue.plot(kind="bar",label="")
plt.title("yearly_revenue")
plt.show()
```

```python
# Monthly Revenue
```

```python
monthly_revenue=marge_data.groupby("month")["TotalAmount"].sum()

monthly_revenue.plot(kind="line",marker="o")
plt.title("manthly_revenue")
plt.show()
```

```python
# Revenue by Country
```

```python
country_revenue=marge_data.groupby("Country")["TotalAmount"].sum()

country_revenue.plot(kind="bar")
plt.title("cuntry by revenue")
plt.show()
```

```python
#Top Customers
```

```python
top_10_customer=marge_data.groupby("CustomerName")["TotalAmount"].sum().sort_values(ascending=False).head(10)

top_10_customer.plot(kind="barh")
plt.title("revenue_by_top_ten_customers")
plt.show()
```

```python
#Orders Status
```

```python
order_status = company["OrderStatus"].value_counts()

order_status.plot(kind="pie", autopct="%1.1f%%")
plt.title("Order Status")
plt.show()
```

```python
#Profit by Product
```

```python
profit_product = company.groupby("ProductName")["profit"].sum().sort_values(ascending=False)

profit_product.head(10).plot(kind="bar")
plt.title("Top Profit Products")
plt.show()
```
